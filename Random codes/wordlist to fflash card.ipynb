```python
import re

def clean_vocab_file(input_path, output_path):
    cleaned_lines = []

    with open(input_path, "r", encoding="utf-8") as f:
        for line in f:
            original = line.strip()

            # --- 1. Remove the IELTS vocabulary line ---
            # Matches: IELTS vocabulary Band +8 "anything"
            if re.match(r'IELTS vocabulary Band \+8 ".*"', original):
                continue

            # --- 2. Split multiple definitions on one line ---
            # Example: "Persuade = get s to agree Convinced = sure of S"
            parts = re.split(r'(?<=\S)\s+(?=[A-Za-z]+\s*=)', original)

            for p in parts:
                p = p.strip()
                if p:
                    cleaned_lines.append(p)

    # Write cleaned output
    with open(output_path, "w", encoding="utf-8") as f:
        for line in cleaned_lines:
            f.write(line + "\n")

    print("Cleaning complete. Output saved to:", output_path)


# Example usage:
inpuT = "C:/Users/razul/Documents/Github/Portfolio/Random codes/wordlist.txt"
clean_vocab_file(inpuT, "output.txt")
```
